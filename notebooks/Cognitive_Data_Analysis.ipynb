```python
final_cognitive = pd.read_csv('/content/drive/MyDrive/ADNI_Project/Cognitive_Data/final_cognitive_702.csv')
```

```python
import matplotlib.pyplot as plt

final_cognitive["DIAGNOSIS"].value_counts().sort_index().plot(kind="bar")

plt.xticks(
    [0,1,2],
    ["CN","MCI","AD"]
)

plt.ylabel("Participants")
plt.title("Diagnosis Distribution in Final Matched Cohort")

plt.show()
```

```python
plt.figure(figsize=(6,4))

plt.hist(final_cognitive["AGE"], bins=20)

plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution")

plt.show()
```

```python
final_cognitive["GENDER"].value_counts()
```

```python
final_cognitive["GENDER"].value_counts().sort_index().plot(
    kind="bar"
)

plt.xticks(
    [0,1],
    ["Male","Female"],
    rotation=0
)

plt.ylabel("Participants")
plt.title("Gender Distribution")

plt.show()
```

```python
final_cognitive.describe()
```

```python
import seaborn as sns

plt.figure(figsize=(12,8))

sns.heatmap(
    final_cognitive.corr(numeric_only=True),
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Matrix")

plt.show()
```

```python
import seaborn as sns

sns.boxplot(
    x="DIAGNOSIS",
    y="MMSCORE",
    data=final_cognitive
)

plt.xticks(
    [0,1,2],
    ["CN","MCI","AD"]
)

plt.title("MMSE Score by Diagnosis")
plt.show()
```
